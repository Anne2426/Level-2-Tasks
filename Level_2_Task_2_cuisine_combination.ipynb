```python
import pandas as pd

df = pd.read_csv('/content/Dataset .csv')

print('First 5 rows of the DataFrame:')
print(df.head())

print('\nDataFrame Information:')
df.info()
```

## Analyze Restaurants per City



```python
restaurants_per_city = df.groupby('City')['Restaurant ID'].nunique().sort_values(ascending=False)

print('Number of restaurants per city (Top 10):')
print(restaurants_per_city.head(10))

highest_restaurants_city = restaurants_per_city.index[0]
highest_restaurants_count = restaurants_per_city.iloc[0]
print(f"\nCity with the highest number of restaurants: {highest_restaurants_city} ({highest_restaurants_count} restaurants)")
```

## Calculate Average Rating per City

### Subtask:
Group the dataset by city and calculate the average rating for restaurants in each city to determine the city with the highest average rating.


**Reasoning**:
To calculate the average rating per city, I will group the DataFrame by 'City', compute the mean of 'Aggregate rating', and then sort the results to identify the city with the highest average rating.



```python
average_rating_per_city = df.groupby('City')['Aggregate rating'].mean().sort_values(ascending=False)

print('Average rating per city (Top 10):')
print(average_rating_per_city.head(10))

highest_avg_rating_city = average_rating_per_city.index[0]
highest_avg_rating_value = average_rating_per_city.iloc[0]
print(f"\nCity with the highest average rating: {highest_avg_rating_city} ({highest_avg_rating_value:.2f})")
```

## Identify Most Common Cuisine Combinations

### Subtask:
Extract and count the occurrences of cuisine combinations to find the most popular ones, handling the 'Cuisines' column which may contain multiple cuisines separated by commas.


**Reasoning**:
To identify the most common cuisine combinations, I will first handle missing values in the 'Cuisines' column, then split the cuisine strings into lists, flatten these lists, and finally count the occurrences of each cuisine to find the top 10.



```python
import collections

# 1. Fill any missing values in the 'Cuisines' column with an empty string
df['Cuisines'].fillna('', inplace=True)

# 2. Split the 'Cuisines' column entries by the comma and space (', ') delimiter into lists of individual cuisines.
#    Strip any leading/trailing whitespace from each cuisine.
df['Cuisine_List'] = df['Cuisines'].apply(lambda x: [c.strip() for c in x.split(',') if c.strip()])

# 3. Create a single, flattened list of all individual cuisines from the Cuisine_List column.
all_cuisines = [cuisine for sublist in df['Cuisine_List'] for cuisine in sublist]

# 4. Count the occurrences of each unique cuisine in the flattened list.
cuisine_counts = collections.Counter(all_cuisines)

# Convert to pandas Series for easy sorting and display
cuisine_series = pd.Series(cuisine_counts).sort_values(ascending=False)

# 5. Print the top 10 most common cuisines.
print('Top 10 most common cuisines:')
print(cuisine_series.head(10))
```

**Reasoning**:
The previous code produced a `FutureWarning` regarding the use of `inplace=True` with chained assignment. To resolve this, I will modify the `fillna` call to assign the result back to the 'Cuisines' column, ensuring the operation is performed on the original DataFrame without triggering the warning.



```python
import collections

# 1. Fill any missing values in the 'Cuisines' column with an empty string
df['Cuisines'] = df['Cuisines'].fillna('')

# 2. Split the 'Cuisines' column entries by the comma and space (', ') delimiter into lists of individual cuisines.
#    Strip any leading/trailing whitespace from each cuisine.
df['Cuisine_List'] = df['Cuisines'].apply(lambda x: [c.strip() for c in x.split(',') if c.strip()])

# 3. Create a single, flattened list of all individual cuisines from the Cuisine_List column.
all_cuisines = [cuisine for sublist in df['Cuisine_List'] for cuisine in sublist]

# 4. Count the occurrences of each unique cuisine in the flattened list.
cuisine_counts = collections.Counter(all_cuisines);

# Convert to pandas Series for easy sorting and display
cuisine_series = pd.Series(cuisine_counts).sort_values(ascending=False)

# 5. Print the top 10 most common cuisines.
print('Top 10 most common cuisines:')
print(cuisine_series.head(10))
```

## Calculate Average Ratings per Cuisine Combination

### Subtask:
Calculate the average rating for each unique cuisine combination and identify the top-rated combinations.

**Reasoning**:
To calculate the average rating for each cuisine, I will iterate through each restaurant, extract its individual cuisines, and then store its aggregate rating for each of those cuisines. Finally, I will compute the mean rating for each cuisine and sort them to find the top-rated combinations.



```python
from collections import defaultdict

# Initialize a defaultdict to store ratings for each cuisine
cuisine_ratings = defaultdict(list)

# Iterate through each row to collect ratings for each cuisine
for index, row in df.iterrows():
    rating = row['Aggregate rating']
    for cuisine in row['Cuisine_List']:
        cuisine_ratings[cuisine].append(rating)

# Calculate the average rating for each cuisine
average_cuisine_ratings = {cuisine: sum(ratings) / len(ratings) for cuisine, ratings in cuisine_ratings.items()}

# Convert to pandas Series for easy sorting and display
average_cuisine_series = pd.Series(average_cuisine_ratings).sort_values(ascending=False)

# Print the top 10 cuisines by average rating
print('Top 10 cuisines by average rating:')
print(average_cuisine_series.head(10))
```

## Visualize Cuisine Combination Analysis

### Subtask:
Generate a plot to visualize the most common cuisine combinations and their corresponding average ratings. Ensure clear labels, titles, and legends for better understanding.


**Reasoning**:
To visualize the most common cuisine combinations and their average ratings, I will create two bar plots side-by-side: one for the top 10 most common cuisines and another for the top 10 cuisines by average rating, ensuring all plots have appropriate labels, titles, and legends for clarity, and handling potential `FutureWarning` from seaborn.



```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create a figure with two subplots side-by-side
fig, axes = plt.subplots(1, 2, figsize=(18, 6))

# Plot 1: Top 10 Most Common Cuisines
sns.barplot(x=cuisine_series.head(10).index, y=cuisine_series.head(10).values, ax=axes[0], palette='viridis', hue=cuisine_series.head(10).index, legend=False)
axes[0].set_title('Top 10 Most Common Cuisines')
axes[0].set_xlabel('Cuisine')
axes[0].set_ylabel('Number of Restaurants')
axes[0].tick_params(axis='x', rotation=45)

# Plot 2: Top 10 Cuisines by Average Rating
sns.barplot(x=average_cuisine_series.head(10).index, y=average_cuisine_series.head(10).values, ax=axes[1], palette='magma', hue=average_cuisine_series.head(10).index, legend=False)
axes[1].set_title('Top 10 Cuisines by Average Rating')
axes[1].set_xlabel('Cuisine')
axes[1].set_ylabel('Average Rating')
axes[1].tick_params(axis='x', rotation=45)

# Add a main title to the entire figure
fig.suptitle('Cuisine Analysis: Popularity vs. Average Rating', fontsize=16);

# Ensure all elements fit within the figure without overlapping
plt.tight_layout(rect=[0, 0.03, 1, 0.95]) # Adjust rect to prevent title overlap

# Display the plot
plt.show()
```
